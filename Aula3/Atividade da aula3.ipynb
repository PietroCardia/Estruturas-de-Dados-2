```
class Paciente:
    def __init__(self, nome, idade, prioritario=False):
        self.nome = nome
        self.idade = idade
        self.prioritario = prioritario

    def __repr__(self):
        tipo = "Prioritário" if self.prioritario else "Comum"
        return f"{self.nome} - {self.idade} anos - {tipo}"


class No:
    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


class FilaAtendimento:
    def __init__(self):
        self.inicio = None
        self.fim = None
        self.tamanho = 0

    def esta_vazia(self):
        return self.inicio is None

    def adicionar(self, paciente):
        novo = No(paciente)

        if self.esta_vazia():
            self.inicio = novo
            self.fim = novo
        elif paciente.prioritario:
            if not self.inicio.paciente.prioritario:
                novo.proximo = self.inicio
                self.inicio = novo
            else:
                atual = self.inicio

                while atual.proximo and atual.proximo.paciente.prioritario:
                    atual = atual.proximo

                novo.proximo = atual.proximo
                atual.proximo = novo

                if novo.proximo is None:
                    self.fim = novo
        else:
            self.fim.proximo = novo
            self.fim = novo

        self.tamanho += 1

    def atender(self):
        if self.esta_vazia():
            return None

        paciente = self.inicio.paciente
        self.inicio = self.inicio.proximo
        self.tamanho -= 1

        if self.inicio is None:
            self.fim = None

        return paciente

    def listar(self):
        if self.esta_vazia():
            print("Fila vazia.")
            return

        atual = self.inicio
        posicao = 1

        while atual:
            print(f"{posicao} - {atual.paciente}")
            atual = atual.proximo
            posicao += 1


def menu():
    fila = FilaAtendimento()

    while True:
        print("\n=== FILA DE ATENDIMENTO ===")
        print("1 - Cadastrar paciente")
        print("2 - Atender paciente")
        print("3 - Listar fila")
        print("4 - Quantidade de pacientes")
        print("0 - Sair")

        opcao = input("Escolha uma opção: ")

        if opcao == "1":
            nome = input("Nome: ")
            try:
                idade = int(input("Idade: "))
            except ValueError:
                print("Idade inválida.")
                continue

            if idade >= 60:
                prioritario = True
                print("Paciente prioritário por idade.")
            else:
                resposta = input("Possui outra prioridade? (s/n): ").lower()
                prioritario = resposta == "s"

            paciente = Paciente(nome, idade, prioritario)
            fila.adicionar(paciente)
            print("Paciente cadastrado com sucesso.")

        elif opcao == "2":
            paciente = fila.atender()

            if paciente:
                print(f"Atendendo: {paciente}")
            else:
                print("Não há pacientes na fila.")

        elif opcao == "3":
            fila.listar()

        elif opcao == "4":
            print(f"Quantidade de pacientes: {fila.tamanho}")

        elif opcao == "0":
            print("Sistema encerrado.")
            break
        else:
            print("Opção inválida.")

if __name__ == "__main__":
    menu()
```
