**Exercicio- Simulador Desfazer**

```python
# Classe Desfazer que inicializa com 2 listas para simular uma pilha.
class Desfazer:
    def __init__(self):
        self.editorTexto = []
        self.pilhaAcoes = [] # Pilha para conseguir fazer a operação inversa da ultima operação feita

# Metodo para adicionar um texto, e salva nas ações da pilha com sua operação e valor
    def digitar(self):
        texto = input("Digite aqui: ")
        self.editorTexto.append(texto)
        self.pilhaAcoes.append({'operacao': "append", 'valor': texto })
        print(f"Conteúdo atual: {self.editorTexto}")

# Metodo para apagar um texto, e salva nas ações da pilha com sua operação e valor
    def apagar(self):
        if self.editorTexto:
            popped_value = self.editorTexto.pop()
            self.pilhaAcoes.append({'operacao': "pop", 'valor': popped_value})
            print(f"Conteúdo atual: {self.editorTexto}")
        else:
            print("Editor de texto está vazio.")

# Método para desfazer a ultima operação feita, acessando os dados salvos na pilha pilhaAcoes e verificando a operação e o valor
    def desfazer(self):
        if not self.pilhaAcoes:
            print("Não há mais ações para desfazer.")
            return

        ultima_acao = self.pilhaAcoes.pop()

        if ultima_acao['operacao'] == 'append':
            if self.editorTexto:
                self.editorTexto.pop()
            else:
                print("Erro: editorTexto está vazio, mas uma ação de 'append' foi registrada.")
        elif ultima_acao['operacao'] == 'pop':
            self.editorTexto.append(ultima_acao['valor'])

        print(f"Conteúdo atual após desfazer: {self.editorTexto}")

if __name__ == "__main__":
    editor = Desfazer()
    editor.digitar()
    editor.digitar()
    editor.apagar()
    editor.desfazer() # Adicionado para testar o desfazer
    editor.desfazer() # Adicionado para testar o desfazer novamente
```

**Exercicio 2- Sistema de Impressão**

```python
from collections import deque

class SistemaImpressao:
  def __init__(self):
    # Inicializa a fila como um atributo da instância
    self.filaImpressao = deque()

  def adicionarFila(self):
    adicionar = input("Digite aqui:")
    # Usa o atributo da instância para adicionar à fila
    self.filaImpressao.append(adicionar)
    print(f"Fila atual: {self.filaImpressao}")

  def imprimir(self):
    if self.filaImpressao:
      # Usa o atributo da instância para remover da fila
      item = self.filaImpressao.popleft()
      print(f"Imprimindo: {item}")
      print(f"Fila atual após impressão: {self.filaImpressao}")
    else:
      print("A fila de impressão está vazia.")

  def mostrarFila(self):
    # Usa o atributo da instância para mostrar a fila
    print(f"Fila atual: {self.filaImpressao}")

if __name__ == "__main__":
  fila = SistemaImpressao()
  fila.adicionarFila()
  fila.adicionarFila()
  fila.mostrarFila()
  fila.imprimir()
  fila.mostrarFila()
  fila.imprimir()
  fila.mostrarFila()
```

**Desafio Master** Triagem Hospitalar

```python
from collections import deque

# Classe para criar a fila hospitalar usando a estrutura fila
class Hospitalar:
  def __init__(self):
     self.filaHospitalar = deque()

# Metodo para adicionar pacientes, adicionando no final da fila, que é seu comportamento padrão
  def adicionarPaciente(self):
    paciente = input("Nome do paciente: ")
    self.filaHospitalar.append(paciente)
    print(self.filaHospitalar)

# Adicionando paciente prioritario, nesse caso ele adiciona no inicio da fila com intuito de sair logo dela
  def adicionarPacientePrioritario(self):
    paciente_prioritario = input("Nome do paciente prioritário: ")
    self.filaHospitalar.appendleft(paciente_prioritario)
    print(self.filaHospitalar)

# Metodo para remover o paciente, aqui ele verifica se a fila esta vazia, se não estiver ele retira do inicio
  def removerPaciente(self):
    if self.filaHospitalar:
      paciente_removido = self.filaHospitalar.popleft()
      print(f"Paciente removido: {paciente_removido}")
    else:
      print("A fila está vazia.")
    print(self.filaHospitalar)

# Metodo que mostra a fila
  def mostrarFila(self):
    print(self.filaHospitalar)

# Testes do uso da clase, instanciando ela e usando os metodos
if __name__ == "__main__":
  filaHospitalar = Hospitalar()
  filaHospitalar.adicionarPaciente()
  filaHospitalar.adicionarPaciente()
  filaHospitalar.adicionarPacientePrioritario()
  filaHospitalar.removerPaciente()
  filaHospitalar.mostrarFila()
```
